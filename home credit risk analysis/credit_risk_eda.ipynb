```python
import pandas as pd
```

```python
df = pd.read_csv('./data/main_test.csv')

# checking the strings into lower case
df.columns = df.columns.str.lower()
```

```python
# converting the days_birth into the age
df['age'] = (df['days_birth'] * (-1)) // 365

# converting the days_employed into the 
df['year_employed'] = (df['days_employed']) * (-1) // 365
```

```python
# the negative years are handled 
df.loc[df['year_employed']<0 , 'year_employed'] = 0

# replacing the features with the newly created features 
df['days_birth'] = df['age']
df['days_employed'] = df['year_employed']

df.drop(columns=['age','year_employed'],inplace=True)
df.rename(columns={'days_birth':'age',
           'days_employed':'year_employed'},inplace=True)
```

```python
# creating the age groups 
df['age_group'] = df['age'] 
df.loc[df['age']<30,'age_group'] = '<30'
df.loc[(df['age']>=30) & (df['age']<40),'age_group'] = '30-40'
df.loc[(df['age']>=40) & (df['age']<50),'age_group'] = '40-50'
df.loc[df['age']>=50,'age_group'] = '50+'


df['age'] = df['age_group']
df.drop(columns= 'age_group',inplace=True)
df.rename(columns = {'age':'age_group'},inplace=True)
```

```python
# calculating the default rate for each age_group 
df.groupby('age_group')['target'].mean()
```

```python
default_rate_without_app_rej = df.loc[df['total_rejections']==0, 'target'].mean()
default_rate_with_app_rej = df.loc[df['total_rejections']>0, 'target'].mean()

print(f"default_rate_with_app_rej: {default_rate_with_app_rej},default_rate_without_app_rej {default_rate_without_app_rej}")
```
